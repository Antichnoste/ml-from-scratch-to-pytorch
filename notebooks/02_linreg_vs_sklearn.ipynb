# 02. Линейная регрессия: с нуля vs sklearn

Обучаем линейную регрессию через градиентный спуск (реализация в `src/part1_from_scratch/linear_regression.py`)
и сравниваем со `sklearn.linear_model.LinearRegression` на датасете Diabetes.


```python
import sys
sys.path.append('../src/part1_from_scratch')

import matplotlib.pyplot as plt

from linear_regression import run_comparison
```

```python
results = run_comparison()
```

## Кривая обучения (loss по эпохам)


```python
plt.figure(figsize=(8, 5))
plt.plot(results['history'])
plt.xlabel('Эпоха')
plt.ylabel('MSE (train)')
plt.title('Сходимость градиентного спуска — линейная регрессия')
plt.grid(alpha=0.3)
plt.savefig('../results/linreg_loss_curve.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Вывод

Loss монотонно убывает и выходит на плато — признак того, что `lr` подобран корректно и
градиентный спуск действительно сходится к оптимуму, близкому к аналитическому решению sklearn (OLS).
Разница в MSE между нашей реализацией и sklearn минимальна (см. вывод ячейки выше).
